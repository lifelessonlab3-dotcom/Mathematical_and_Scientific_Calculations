```python
7) Compound Interest:
Write a Python program to calculate compound interest. Use the formula:
Where:
A= total amount
P = principal amount
r = annual interest rate (decimal)
n = number of times interest is compounded per year
t = time in years
Take P, r, n, and t as inputs from the user.
```

```python
principal = float(input("Enter the principal amount: "))
rate = float(input("Enter the annual interest rate (decimal): "))
n = int(input("Enter the number of times interest is compounded per year: "))
time = float(input("Enter the time in years: "))

amount = principal * (1 + rate / n) ** (n * time)

compound_interest = amount - principal

print("Total Amount =", amount)
print("Compound Interest =", compound_interest)
```

```python
19) Work Done by a Force:
Write a Python program to calculate the work done by a force. Use the formula:

where:
W = work done
f= force
d = distance
theta = angle between force and direction of movement (in degrees)
Take f, d, and θ as inputs from the user.
```

```python
import math

force = float(input("Enter the force: "))
distance = float(input("Enter the distance: "))
theta = float(input("Enter the angle in degrees: "))

# Convert degrees to radians
theta_radians = math.radians(theta)

work = force * distance * math.cos(theta_radians)

print("Work Done =", work, "Joules")
```

```python
20) Heat Transfer:
Write a Python program to calculate the amount of heat transferred. Use the formula:

Q = mc (∆T)

where:
Q= heat transfer
m = mass
c = specific heat capacity
ΔT
Take m, c, T as inputs from the user.
```

```python
mass = float(input("Enter the mass: "))
specific_heat = float(input("Enter the specific heat capacity: "))
initial_temp = float(input("Enter the initial temperature: "))
final_temp = float(input("Enter the final temperature: "))

delta_t = final_temp - initial_temp

heat = mass * specific_heat * delta_t

print("Heat Transfer =", heat, "Joules")
```
